```python
import math
import random

def objective_function(x):
    """The function we want to minimize."""
    return x**2

def simulated_annealing(bounds, iterations, temp):
    # Start with a random initial point
    current_state = random.uniform(bounds[0], bounds[1])
    current_eval = objective_function(current_state)
    
    best_state, best_eval = current_state, current_eval
    
    for i in range(iterations):
        # 1. Generate a candidate (neighbor)
        candidate = current_state + random.uniform(-1, 1)
        candidate = max(bounds[0], min(bounds[1], candidate)) # Stay in bounds
        candidate_eval = objective_function(candidate)
        
        # 2. Check if we should accept the candidate
        if candidate_eval < current_eval:
            # Better solution: always accept
            current_state, current_eval = candidate, candidate_eval
            if candidate_eval < best_eval:
                best_state, best_eval = candidate, candidate_eval
        else:
            # Worse solution: accept with a probability (Metropolis Criterion)
            diff = candidate_eval - current_eval
            # Temperature decreases over time (Linear Cooling)
            t = temp / float(i + 1)
            metropolis = math.exp(-diff / t)
            if random.random() < metropolis:
                current_state, current_eval = candidate, candidate_eval
                
    return best_state, best_eval

# Configuration
bounds = [-10, 10]
total_iterations = 1000
initial_temp = 10

# Execution
best_x, score = simulated_annealing(bounds, total_iterations, initial_temp)
print(f"Global Minimum found at x = {best_x:.4f} with f(x) = {score:.4f}")
```

```python
import math
import random

def objective_function(x):
    """The function we want to minimize (f(x) = x^2)."""
    return x**2

def simulated_annealing():
    print("--- Simulated Annealing Configuration ---")
    try:
        # Taking inputs from the user
        temp = float(input("Enter Initial Temperature (e.g., 1000): "))
        cooling_rate = float(input("Enter Cooling Rate (e.g., 0.95): "))
        iterations = int(input("Enter Number of Iterations (e.g., 500): "))
        
        # Start with a random initial state within a range
        current_x = random.uniform(-100, 100)
        current_eval = objective_function(current_x)
        
        best_x = current_x
        best_eval = current_eval

        print(f"\nStarting at x = {current_x:.4f}, f(x) = {current_eval:.4f}")

        for i in range(iterations):
            # 1. Generate a neighbor (a small step away)
            neighbor_x = current_x + random.uniform(-1, 1)
            neighbor_eval = objective_function(neighbor_x)
            
            # 2. Calculate the difference in cost
            diff = neighbor_eval - current_eval
            
            # 3. Metropolis Criterion
            # If neighbor is better, move there. 
            # If worse, move there with a probability that decreases as temp drops.
            if diff < 0 or random.random() < math.exp(-diff / temp):
                current_x = neighbor_x
                current_eval = neighbor_eval
                
                # Keep track of the best solution found so far
                if current_eval < best_eval:
                    best_x = current_x
                    best_eval = current_eval
            
            # 4. Decrease temperature
            temp *= cooling_rate

        print("\n--- Results ---")
        print(f"Best x found: {best_x:.6f}")
        print(f"Objective Function Value: {best_eval:.6f}")

    except ValueError:
        print("Invalid input! Please enter numbers for temperature, rate, and iterations.")
    except ZeroDivisionError:
        print("Temperature reached zero too quickly. Try a slower cooling rate.")

if __name__ == "__main__":
    simulated_annealing()
```
