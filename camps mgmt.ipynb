```python
students = set()
teachers = set()

def show_all_records():
    """Displays all stored records using string operations for formatting."""
    print("\n" + "="*30)
    print("      CAMPUS DIRECTORY")
    print("="*30)
    
    print("\n[STUDENTS]")
    if not students: print("No student records.")
    for s in students:
        # Use split to break the string into a list
        id_val, name_val, course_val = s.split(":")
        # Use upper for ID and Course, and capitalize for the name
        print(f"ID: {id_val.upper()} | Name: {name_val.title()} | Course: {course_val.upper()}")

    print("\n[TEACHERS]")
    if not teachers: print("No teacher records.")
    for t in teachers:
        id_val, name_val, course_val = t.split(":")
        print(f"ID: {id_val.upper()} | Name: {name_val.title()} | Course: {course_val.upper()}")
    print("="*30)

while True:
    print("\n1. Add Student  2. Add Teacher  3. Show All Records  4. Search Course  5. Exit")
    choice = input("Select Option: ")

    if choice in ['1', '2']:
        # Taking inputs separately
        uid = input("Enter ID: ")
        name = input("Enter Name: ")
        course = input("Enter Course ID: ")
        
        # Using lower() to standardize data before adding to set
        record = f"{uid.lower()}:{name.lower()}:{course.lower()}"
        
        if choice == '1':
            students.add(record)
            print("Student added successfully.")
        else:
            teachers.add(record)
            print("Teacher added successfully.")

    elif choice == '3':
        show_all_records()

    elif choice == '4':
        query = input("Enter Course ID to search: ").lower()
        print(f"\nResults for: {query.upper()}")
        
        # Using membership operator 'in'
        for s in students:
            if query in s:
                print(f"Student: {s.split(':')[1].title()}")
        for t in teachers:
            if query in t:
                print(f"Teacher: {t.split(':')[1].title()}")

    elif choice == '5':
        break
```

```python
import tkinter as tk
from tkinter import messagebox

# Storage sets
students = set()
teachers = set()
all_ids = set()

# Master Course List for recommendations
MASTER_COURSES = {"cs101", "cs102", "math201", "phy101", "eng301", "bio102"}

def extract_courses(data_set):
    """Helper: Uses split() and set comprehension to get a clean set of course IDs."""
    return {item.split(":")[2] for item in data_set}

def add_record(role):
    uid = ent_id.get().lower().strip()
    name = ent_name.get().lower().strip()
    course = ent_course.get().lower().strip()

    if not uid or not name or not course:
        messagebox.showwarning("Input Error", "All fields are required!")
        return
    
    # Membership operator for Unique ID validation
    if uid in all_ids:
        messagebox.showerror("Error", f"ID '{uid.upper()}' is already taken!")
        return

    record = f"{uid}:{name}:{course}"
    if role == "Student":
        students.add(record)
    else:
        teachers.add(record)
    
    all_ids.add(uid)
    messagebox.showinfo("Success", f"{role} added successfully!")
    clear_entries()

def show_all_records():
    display.delete("1.0", tk.END)
    display.insert(tk.END, f"{'ROLE':<10} | {'ID':<10} | {'NAME':<15} | {'COURSE'}\n")
    display.insert(tk.END, "-"*55 + "\n")
    for category, data_set in [("Student", students), ("Teacher", teachers)]:
        for item in data_set:
            u, n, c = item.split(":")
            display.insert(tk.END, f"{category:<10} | {u.upper():<10} | {n.title():<15} | {c.upper()}\n")

def search_course():
    query = ent_course.get().lower().strip()
    if not query:
        messagebox.showwarning("Search", "Enter a Course ID in the Course field.")
        return
    display.delete("1.0", tk.END)
    display.insert(tk.END, f"--- Results for {query.upper()} ---\n")
    for s in students:
        if query in s: display.insert(tk.END, f"Student: {s.split(':')[1].title()}\n")
    for t in teachers:
        if query in t: display.insert(tk.END, f"Teacher: {t.split(':')[1].title()}\n")

def show_analytics():
    """Uses UNION and INTERSECTION set operations."""
    s_courses = extract_courses(students)
    t_courses = extract_courses(teachers)
    
    all_active = s_courses | t_courses  # UNION
    both_present = s_courses & t_courses # INTERSECTION
    
    display.delete("1.0", tk.END)
    display.insert(tk.END, f"Total Active Courses (Union): {len(all_active)}\n")
    display.insert(tk.END, f"Courses with Teacher & Student (Intersection): {list(both_present)}\n")

def recommend_new_courses():
    """Uses DIFFERENCE set operation."""
    s_courses = extract_courses(students)
    # Recommended = Courses in Master list but NOT yet in Student set
    recommended = MASTER_COURSES - s_courses # DIFFERENCE
    
    display.delete("1.0", tk.END)
    display.insert(tk.END, "--- RECOMMENDED FOR STUDENTS ---\n")
    for c in recommended:
        display.insert(tk.END, f"- {c.upper()}\n")

def clear_entries():
    ent_id.delete(0, tk.END); ent_name.delete(0, tk.END); ent_course.delete(0, tk.END)

# --- GUI Setup ---
root = tk.Tk()
root.title("Campus Management System Pro")
root.geometry("550x550")

# Entry Widgets
tk.Label(root, text="ID:").grid(row=0, column=0, pady=2)
ent_id = tk.Entry(root); ent_id.grid(row=0, column=1)
tk.Label(root, text="Name:").grid(row=1, column=0, pady=2)
ent_name = tk.Entry(root); ent_name.grid(row=1, column=1)
tk.Label(root, text="Course:").grid(row=2, column=0, pady=2)
ent_course = tk.Entry(root); ent_course.grid(row=2, column=1)

# Buttons - Row 3 & 4
tk.Button(root, text="Add Student", command=lambda: add_record("Student"), width=15).grid(row=3, column=0, pady=5)
tk.Button(root, text="Add Teacher", command=lambda: add_record("Teacher"), width=15).grid(row=3, column=1)
tk.Button(root, text="Show All", command=show_all_records, width=15).grid(row=4, column=0, pady=5)
tk.Button(root, text="Search Course", command=search_course, width=15).grid(row=4, column=1)

# New Set Operation Buttons - Row 5
tk.Button(root, text="Analytics (U/I)", command=show_analytics, width=15, fg="blue").grid(row=5, column=0, pady=5)
tk.Button(root, text="Recommend (-)", command=recommend_new_courses, width=15, fg="green").grid(row=5, column=1)

# Display Area
display = tk.Text(root, height=15, width=65)
display.grid(row=6, column=0, columnspan=2, padx=10, pady=10)

root.mainloop()
```
