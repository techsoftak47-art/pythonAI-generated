```python
import csv

def list_chennai_data(filename):
    print("Name and Phone Number for Chennai City:")
    print("-" * 40)
    
    try:
        with open(filename, mode='r', newline='', encoding='utf-8') as file:
            # Use DictReader to handle header columns directly
            reader = csv.DictReader(file)
            
            # Loop through rows, assuming columns: name, city, phone
            for row in reader:

                #print(row['contact'])
                # Compare city, ignoring case
                if row['city'].strip().lower() == 'chennai':
                    print(f"Name: {row['name']}")
                    print(f"Phone: {row['contact']}")
    except FileNotFoundError:
        print("Error: The file does not exist.")
    except KeyError:
        print("Error: Ensure CSV has 'name', 'city', and 'contact' columns.")

# Example usage assuming the file exists
filename = r"C:\Users\anju\Downloads\land.csv"
list_chennai_data(filename)
```

```python
# 1. Using Tuples for Immutable Records
# Format: (Name, Specialization/Issue)
doctor1 = ("Dr. Smith", "Cardiology")
doctor2 = ("Dr. Jones", "Neurology")
patient1 = ("Alice", "Heart Checkup")
patient2 = ("Bob", "Headache")

# 2. Using Lists to Store Multiple Records
doctors_list = [doctor1, doctor2]
patients_list = [patient1, patient2]

# 3. Using Dictionary to Map Patient to Doctor (Simple Relationship)
clinic_data = {
    "Doctors": doctors_list,
    "Patients": patients_list
}






# --- Output Scenario ---
print("--- Hospital Management System ---")
'''
dictnm['key']==value
'''
print(f"Doctor on duty: {clinic_data['Doctors'][0][0]} ({clinic_data['Doctors'][0][1]})")
print(f"Patient registered: {clinic_data['Patients'][0][0]} for {clinic_data['Patients'][0][1]}")

print("\nFull System Records (List of Tuples):")
for key, value in clinic_data.items():
    print(f"{key}: {value}")
```
