# Finding Maximum Value in a List
# Route Optimization (Travelling Salesman)
# Campus Navigation

```python
import random

# --- ALGORITHM FUNCTIONS ---

def hill_climbing_list(data):
    #max values
    if not data: return None
    current_idx = len(data) // 2  # Start in the middle
    while True:
        current_val = data[current_idx]
        left_val = data[current_idx - 1] if current_idx > 0 else float('-inf')
        right_val = data[current_idx + 1] if current_idx < len(data) - 1 else float('-inf')

        if left_val > current_val and left_val >= right_val:
            current_idx -= 1
        elif right_val > current_val and right_val >= left_val:
            current_idx += 1
        else:
            return data[current_idx], current_idx

def get_route_distance(route, dist_matrix):
    total = 0
    for i in range(len(route) - 1):
        total += dist_matrix[route[i]][route[i+1]]
    return total + dist_matrix[route[-1]][route[0]]

def optimize_route(nodes, dist_matrix):
    current_route = list(nodes)
    random.shuffle(current_route)
    current_dist = get_route_distance(current_route, dist_matrix)
    
    for _ in range(1000):
        a, b = random.sample(range(len(current_route)), 2)
        neighbor = current_route[:]
        neighbor[a], neighbor[b] = neighbor[b], neighbor[a]
        if get_route_distance(neighbor, dist_matrix) < current_dist:
            current_route = neighbor
            current_dist = get_route_distance(neighbor, dist_matrix)
    return current_route, current_dist

def campus_navigation(graph, heuristic, start, goal):
    current_node = start
    path = [current_node]
    while current_node != goal:
        neighbors = graph.get(current_node, {})
        if not neighbors: break
        best_neighbor = min(neighbors.keys(), key=lambda n: heuristic.get(n, float('inf')))
        if heuristic.get(best_neighbor, float('inf')) < heuristic.get(current_node, float('inf')):
            current_node = best_neighbor
            path.append(current_node)
        else:
            print("Reached a local optimum (stuck).")
            break
    return path

# --- MENU INTERFACE ---

def main():
    while True:
        print("\n--- Hill Climbing Menu ---")
        print("1. Find Maximum Value in List")
        print("2. Route Optimization (TSP)")
        print("3. Campus Navigation")
        print("4. Exit")
        choice = input("Enter choice (1-4): ")

        if choice == '1':
            nums = [int(x) for x in input("Enter list of numbers (space separated): ").split()]
            val, idx = hill_climbing_list(nums)
            print(f"Peak found: {val} at index {idx}")

        elif choice == '2':
            n = int(input("Enter number of locations: "))
            matrix = []
            print("Enter distance matrix row by row (space separated values):")
            for i in range(n):
                matrix.append([int(x) for x in input(f"Row {i}: ").split()])
            path, dist = optimize_route(list(range(n)), matrix)
            print(f"Optimized Path: {path}\nTotal Distance: {dist}")

        elif choice == '3':
            # Simplified input for graph/heuristic
            nodes = input("Enter node names (space separated): ").split()
            h = {}
            for node in nodes:
                h[node] = int(input(f"Enter heuristic (distance to goal) for {node}: "))
            
            graph = {}
            print("Enter edges as 'node1 node2' (type 'done' to finish):")
            while True:
                edge = input("> ").split()
                if edge[0].lower() == 'done': break
                u, v = edge[0], edge[1]
                if u not in graph: graph[u] = {}
                graph[u][v] = 1 # simplified weight
            
            start = input("Enter Start node: ")
            goal = input("Enter Goal node: ")
            print(f"Navigation Path: {' -> '.join(campus_navigation(graph, h, start, goal))}")

        elif choice == '4':
            break
        else:
            print("Invalid choice!")

if __name__ == "__main__":
    main()
```
