```python
import heapq

def a_star_algorithm(graph, heuristics, start, goal):
    # The open_list stores tuples: (priority, current_node, path_taken, current_cost)
    # priority (f) = current_cost (g) + heuristic (h)
    open_list = [(heuristics[start], start, [start], 0)]
    
    # Keep track of the best g_score for each node to avoid redundant work
    visited_costs = {start: 0}

    print(f"--- A* Search Trace ---")

    while open_list:
        # Pop the node with the lowest f_score (priority)
        f_score, current_node, path, g_score = heapq.heappop(open_list)

        print(f"Exploring: {current_node:8} | Path: {' -> '.join(path):25} | g: {g_score}, f: {f_score}")

        # Goal check
        if current_node == goal:
            return path, g_score

        # Explore neighbors
        for neighbor, weight in graph.get(current_node, {}).items():
            new_g_score = g_score + weight
            
            # If we found a cheaper way to this neighbor, or it's our first time seeing it
            if neighbor not in visited_costs or new_g_score < visited_costs[neighbor]:
                visited_costs[neighbor] = new_g_score
                f_score = new_g_score + heuristics[neighbor]
                heapq.heappush(open_list, (f_score, neighbor, path + [neighbor], new_g_score))

    return None, float('inf')

# 1. Define the Graph
graph = {
    'gate': {'library': 4, 'Office': 2},
    'library': {'AI Lab': 5},
    'Office': {'canteen': 6, 'AI Lab': 8},
    'canteen': {'AI Lab': 3},
    'AI Lab': {}
}

# 2. Define the Heuristics (estimated distance to AI Lab)
heuristics = {
    'gate': 7,
    'library': 4,
    'Office': 6,
    'canteen': 2,
    'AI Lab': 0
}

# 3. Execute the algorithm
optimal_path, total_cost = a_star_algorithm(graph, heuristics, 'gate', 'AI Lab')

# 4. Print Results
print("\n--- Final Result ---")
if optimal_path:
    print(f"Optimal Path: {' -> '.join(optimal_path)}")
    print(f"Total Cost:   {total_cost}")
else:
    print("No path found.")
```
