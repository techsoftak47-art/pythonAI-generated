```python

import pandas as pd

# 1. Initialize the Student Database
columns = ['Student_ID', 'Name', 'Dept', 'Attendance', 'Marks', 'Placement']
student_df = pd.DataFrame(columns=columns)

def add_student(df):
    """Adds a new student record with strict placement validation."""
    print("\n--- Add New Student ---")
    s_id = input("Enter Student ID: ")
    name = input("Enter Name: ")
    dept = input("Enter Department: ")
    try:
        att = float(input("Enter Attendance (%): "))
        marks = float(input("Enter Marks: "))
    except ValueError:
        print("Error: Attendance and Marks must be numbers.")
        return df
    
    # Validating Placement input
    placement = ""
    while placement not in ['yes', 'no']:
        placement = input("Placement Status (yes/no): ").lower().strip()
        if placement not in ['yes', 'no']:
            print("Invalid input. Please enter 'yes' or 'no'.")

    new_data = pd.DataFrame([[s_id, name, dept, att, marks, placement]], columns=columns)
    df = pd.concat([df, new_data], ignore_index=True)
    print(f"Record for {name} added successfully!")
    return df

def find_average(df):
    """Calculates the average marks of all students."""
    if df.empty:
        print("No records found to calculate average.")
    else:
        avg = df['Marks'].mean()
        print(f"\nAverage Marks across all students: {avg:.2f}")

def search_student(df):
    """Searches for a student by ID or Name."""
    query = input("\nEnter Student ID or Name to search: ")
    result = df[(df['Student_ID'] == query) | (df['Name'].str.contains(query, case=False, na=False))]
    if result.empty:
        print("No student found with those details.")
    else:
        print("\n--- Search Results ---")
        print(result.to_string(index=False))

def performance_analysis(df):
    """Provides a brief analysis based on marks and attendance."""
    if df.empty:
        print("No records available for analysis.")
        return
    
    print("\n--- Student Performance Analysis ---")
    highest = df.loc[df['Marks'].idxmax()]
    print(f"Topper: {highest['Name']} with {highest['Marks']} Marks.")
    
    # Analysis based on attendance thresholds
    low_attendance = df[df['Attendance'] < 75]
    if not low_attendance.empty:
        print(f"Warning: {len(low_attendance)} students have attendance below 75%.")
    
    # Placement vs Marks
    placed_count = len(df[df['Placement'] == 'yes'])
    print(f"Total Placed: {placed_count} / {len(df)}")

# --- Main Menu Loop ---
while True:
    print("\n--- Student Management System ---")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Find Average Marks")
    print("4. Search Student")
    print("5. Performance Analysis")
    print("6. Exit")
    
    choice = input("Select an option (1-6): ")
    
    if choice == '1':
        student_df = add_student(student_df)
    elif choice == '2':
        print(student_df.to_string(index=False)) if not student_df.empty else print("Empty Database.")
    elif choice == '3':
        find_average(student_df)
    elif choice == '4':
        search_student(student_df)
    elif choice == '5':
        performance_analysis(student_df)
    elif choice == '6':
        break
    else:
        print("Invalid choice.")
```
