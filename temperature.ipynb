```python
import tkinter as tk
from tkinter import messagebox
import numpy as np

class SensorSystemGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Intelligent Sensor Monitor")
        self.root.geometry("450x500")

        # Header and Input
        tk.Label(root, text="Sensor Monitoring System", font=('Arial', 14, 'bold')).pack(pady=10)
        tk.Label(root, text="Enter 24 temperature values (comma-separated):").pack()
        
        self.input_field = tk.Entry(root, width=50)
        self.input_field.pack(pady=5)
        # Placeholder data
        self.input_field.insert(0, "22,46,30,55,21,44,32,48,25,29,50,22,41,35,47,38,20,52,43,31,27,49,26,40")

        # Action Button
        self.process_btn = tk.Button(root, text="Run Analysis", command=self.process_data, 
                                     bg="#4CAF50", fg="white", font=('Arial', 10, 'bold'))
        self.process_btn.pack(pady=10)

        # Output Display
        self.results_text = tk.Text(root, height=18, width=55, font=('Courier', 9))
        self.results_text.pack(pady=10, padx=10)

    def process_data(self):
        try:
            # 1. Parse Input
            raw_data = self.input_field.get()
            data_list = [float(x.strip()) for x in raw_data.split(',')]
            
            if len(data_list) != 24:
                messagebox.showwarning("Input Error", f"Expected 24 values, but got {len(data_list)}.")
                return

            # 2. NumPy Processing
            sensor_data = np.array(data_list)
            matrix = sensor_data.reshape(4, 6)
            
            avg_val, max_val, min_val = np.mean(matrix), np.max(matrix), np.min(matrix)
            threshold = 45
            alerts = np.argwhere(matrix > threshold)

            # 3. Update UI
            self.results_text.delete(1.0, tk.END)
            self.results_text.insert(tk.END, f"RESHAPED SENSOR MATRIX (4x6):\n{matrix}\n")
            self.results_text.insert(tk.END, "-"*40 + "\n")
            self.results_text.insert(tk.END, f"AVERAGE: {avg_val:.2f}°C\n")
            self.results_text.insert(tk.END, f"MAXIMUM: {max_val:.2f}°C\n")
            self.results_text.insert(tk.END, f"MINIMUM: {min_val:.2f}°C\n")
            self.results_text.insert(tk.END, "-"*40 + "\n")

            if alerts.size > 0:
                self.results_text.insert(tk.END, f"ALERTS DETECTED (> {threshold}°C):\n", "alert_style")
                for idx in alerts:
                    val = matrix[idx[0], idx[1]]
                    self.results_text.insert(tk.END, f" -> Sensor {idx[0]}, Time {idx[1]}: {val}°C\n")
            else:
                self.results_text.insert(tk.END, "Status: All temperatures within safe limits.")

        except ValueError:
            messagebox.showerror("Error", "Please enter valid numbers only.")

if __name__ == "__main__":
    root = tk.Tk()
    app = SensorSystemGUI(root)
    root.mainloop()
```
